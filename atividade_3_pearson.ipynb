Normalização de vetores e Coeficiente de Correlação de Pearson

## Enunciado

1. Gerar um vetor **A** com 100 números reais aleatórios entre 0 e 100.
2. Gerar um vetor **B** com 100 números reais aleatórios entre 0 e 1.
3. Reescalar os dois vetores para o intervalo entre 5 e 10, usando uma função própria.
4. Calcular o **Coeficiente de Correlação de Pearson** entre os vetores reescalados usando apenas operações básicas do NumPy.
5. Comparar o resultado com `pearsonr` da biblioteca `scipy.stats`.
6. Calcular a correlação nos vetores originais, sem reescala, e comparar os resultados.

## 1. Bibliotecas e configuração

O **NumPy** será utilizado para criar os vetores e realizar as operações matemáticas. A função `pearsonr` do **SciPy** será usada somente no momento da comparação solicitada pelo exercício.

A semente aleatória é definida para que os mesmos vetores possam ser reproduzidos em novas execuções do notebook.

```python
import numpy as np
from scipy.stats import pearsonr

np.random.seed(42)
np.set_printoptions(precision=4, suppress=True)
```

## 2. Geração dos vetores A e B

A função `np.random.rand(100)` gera 100 valores reais no intervalo $[0,1)$. Por isso:

- para **A**, multiplicamos os valores por 100, obtendo números no intervalo $[0,100)$;
- para **B**, usamos diretamente os valores produzidos por `np.random.rand(100)`, mantendo o intervalo $[0,1)$.

```python
A = np.random.rand(100) * 100
B = np.random.rand(100)

print('Vetor A:')
print(A)
print()
print('Vetor B:')
print(B)

print()
print('Dimensão de A:', A.shape)
print('Dimensão de B:', B.shape)
```

## 3. Função própria para reescala Min-Max

Para transformar um valor $x$ do intervalo original para o novo intervalo $[5,10]$, será usado:

$$
x' = 5 + \frac{x - x_{min}}{x_{max} - x_{min}}(10-5)
$$

A função abaixo implementa essa transformação diretamente com operações do NumPy.

```python
def rescale_vector(vector, new_min=5, new_max=10):
    old_min = np.min(vector)
    old_max = np.max(vector)

    scaled_vector = new_min + ((vector - old_min) / (old_max - old_min)) * (new_max - new_min)
    return scaled_vector
```

## 4. Reescala dos vetores para o intervalo [5, 10]

Mesma função aplicada aos dois vetores. Em seguida, verificamos os menores e maiores valores encontrados. Como a reescala usa o mínimo e o máximo observados em cada vetor, o menor valor passa a ser 5 e o maior passa a ser 10.

```python
A_scaled = rescale_vector(A, 5, 10)
B_scaled = rescale_vector(B, 5, 10)

print('A reescalado:')
print(A_scaled)
print()
print('B reescalado:')
print(B_scaled)

print()
print('Intervalo de A reescalado:', np.min(A_scaled), 'a', np.max(A_scaled))
print('Intervalo de B reescalado:', np.min(B_scaled), 'a', np.max(B_scaled))
```

## 5. Implementação do Coeficiente de Correlação de Pearson

Coeficiente da correlação de Pearson para dois vetores $x$ e $y$:

$$
r = \frac{\sum_{i=1}^{N}(x_i-\mu_x)(y_i-\mu_y)}
{\sqrt{\sum_{i=1}^{N}(x_i-\mu_x)^2\sum_{i=1}^{N}(y_i-\mu_y)^2}}
$$

Na implementação, busquei usar apenas cálculos elementares com NumPy: média, subtração, multiplicação, soma, potência e raiz quadrada. Não utilizei `np.corrcoef` nem outra função pronta de correlação.

```python
def pearson_numpy(x, y):
    mean_x = np.mean(x)
    mean_y = np.mean(y)

    diff_x = x - mean_x
    diff_y = y - mean_y

    numerator = np.sum(diff_x * diff_y)
    denominator = np.sqrt(np.sum(diff_x ** 2) * np.sum(diff_y ** 2))

    r = numerator / denominator
    return r
```

## 6. Correlação entre os vetores reescalados

Cálculo  para correlação entre `A_scaled` e `B_scaled` com a função implementada acima.

```python
r_scaled_numpy = pearson_numpy(A_scaled, B_scaled)

print('Correlação de Pearson - implementação com NumPy:')
print(r_scaled_numpy)
```

## 7. Comparação com `scipy.stats.pearsonr`

A função `pearsonr` fornece o coeficiente de correlação de Pearson e também um **p-value** associado ao teste de hipótese de correlação linear nula.

Para responder ao item 1 do exercício, o coeficiente retornado pelo SciPy é comparado com o valor obtido pela implementação anterior.

```python
r_scaled_scipy, p_value = pearsonr(A_scaled, B_scaled)

difference = np.abs(r_scaled_numpy - r_scaled_scipy)

print('Minha implementação:', r_scaled_numpy)
print('scipy.stats.pearsonr:', r_scaled_scipy)
print('Diferença absoluta:', difference)
print('p-value:', p_value)
```

### Interpretação da comparação

Os dois coeficientes devem ser iguais, exceto por possíveis diferenças muito pequenas decorrentes da precisão numérica de ponto flutuante. Isso indica que a implementação direta da fórmula de Pearson está consistente com a função de referência do SciPy.

O `p-value` é apresentado apenas porque faz parte do retorno de `pearsonr`; o foco principal desta atividade é a comparação do **coeficiente $r$**.

## 8. Correlação nos vetores originais, sem reescala

Para responder ao item 2, calculamos novamente a Correlação de Pearson usando os vetores originais `A` e `B`. Depois comparamos esse valor com a correlação obtida após a reescala.

```python
r_original = pearson_numpy(A, B)

difference_original_scaled = np.abs(r_original - r_scaled_numpy)

print('Correlação nos vetores originais:', r_original)
print('Correlação nos vetores reescalados:', r_scaled_numpy)
print('Diferença absoluta:', difference_original_scaled)
```

No cálculo da Correlação de Pearson, os deslocamentos são eliminados pela centralização em relação à média e os fatores positivos de escala são cancelados pela normalização.

Assim, conclui-se que **a reescala foi útil para colocar os vetores A e B na mesma faixa de valores $[5, 10]$, mas não alterou a Correlação de Pearson entre eles**. Neste experimento, tanto antes quanto depois da normalização, os vetores apresentaram praticamente ausência de associação linear.
